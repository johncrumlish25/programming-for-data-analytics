The dataset used for this project was obtained from [Met Éireann – Historical Data](https://www.met.ie/climate/available-data/historical-data). It contains monthly weather observations recorded at the Claremorris Meteorological Station in County Mayo, covering the period from 1950 to 2026. The dataset provides historical climate information that can be used to investigate weather patterns and changes over time.

## Reading in the Data

```python
# Import libraries
import pandas as pd
```

```python
# Read the weather dataset (skipping the intro info)
DATADIR = "project-data/"
FILENAME = "claremorris-monthly-weather.csv"

df = pd.read_csv(DATADIR + FILENAME, skiprows=19)

df.head()
```

```python
# Display info about the dataset
df.info()
```

### Inspecting the Data

The [info()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.info.html) method was used to examine the structure of the dataset. It provides an overview of the number of rows, columns, column names, data types and the number of non-null values in each column.

The output shows that the dataset contains 919 rows and 12 columns. It also identifies which columns contain numerical data and which are currently stored as objects. This is useful for identifying potential data-cleaning requirements before beginning the analysis.

```python
# Display the descriptive stats
df.describe()
```

### Descriptive Statistics

The [describe()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html) method was used to generate descriptive statistics for the numerical columns in the dataset. These statistics provide an overview of the data, including the mean, standard deviation, minimum and maximum.

The results show that the average monthly mean temperature (`meant`) across the dataset is approximately 9.26°C. The average mean maximum temperature (`mnmax`) is approximately 12.99°C, while the average mean minimum temperature (`mnmin`) is approximately 5.53°C.

It is worth noting that the output only includes five of the twelve columns in the dataset. The remaining seven columns are currently stored as objects rather than numerical data types, meaning they have been excluded from these descriptive statistics.
