```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Projected Births by Year

```python
url = "https://raw.githubusercontent.com/andrewbeattycourseware/PFDA-courseware/main/code/data/projectedbirths-cso.csv"

df = pd.read_csv(url)

df.head(10)
```

```python
plt.plot(df["Year"], df["VALUE"])

plt.xlabel("Year")
plt.ylabel("Projected Births")
plt.title("Projected Births in Ireland")

plt.show()
```

### Projected Births by Age & Administrative Counties (2022)

```python
url = "https://raw.githubusercontent.com/andrewbeattycourseware/PFDA-courseware/main/code/data/cso-populationbyage.csv"

age_df = pd.read_csv(url)

age_df.head(10)
```

Used AI past this point to help with getting rid "All Ages" section and improve display numbers on x axis

```python
mayo = age_df[
    (age_df["Administrative Counties"] == "Mayo  County Council") &
    (age_df["Single Year of Age"] != "All ages")
]

mayo.head(10)
```

```python
mayo = age_df[
    (age_df["Administrative Counties"] == "Mayo  County Council") &
    (age_df["Single Year of Age"] != "All ages")
].copy()

mayo["Age"] = mayo["Single Year of Age"].str.extract("(\d+)").astype(int)

mayo = mayo.sort_values("Age")

plt.figure(figsize=(12, 6))

plt.plot(mayo["Age"], mayo["VALUE"])

plt.xlabel("Age")
plt.ylabel("Number of People")
plt.title("Population by Age in Mayo - 2022")

plt.show()
```
