### Pie chart of people's email domains from the CSV file


### Author: John Crumlish

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datadir = "../data/"
filename = datadir+"people-1000.csv"
```

```python
df = pd.read_csv(filename)
df.head(3)
```

```python
# create a new column with just the domain in it
df['domain'] = df['Email'].str.split('@').str[1]
df.head(3)
```

```python
# count how many emails belong to each domain
domain_counts = df['domain'].value_counts()

domain_counts
```

```python
# create a pie chart showing the email domains
domain_counts.plot.pie(autopct="%.1f")

plt.ylabel('')
plt.title('Email Domains')
plt.show()
```
